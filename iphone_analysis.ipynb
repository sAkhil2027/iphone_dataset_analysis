```python
import re

# 1. Extract exact Model (e.g., iPhone 15, iPhone 16 Plus)
df['model'] = df['product_name'].str.extract(r'(iPhone\s+\d+\s*(?:Plus|Pro\s*Max|Pro)?)', expand=False).str.strip()

# 2. Extract Generation (e.g., iPhone 13, iPhone 14, iPhone 15, iPhone 16)
df['generation'] = df['model'].str.extract(r'(iPhone\s+\d+)', expand=False)

# 3. Extract Storage Capacity (Standardized to 128GB, 256GB, etc.)
df['storage'] = df['product_name'].str.extract(r'(\d+\s*(?:GB|TB))', flags=re.IGNORECASE, expand=False).str.upper().str.replace(" ", "")

# 4. Extract Color
color_pattern = r'(Blue|Black|White|Green|Midnight|Starlight|Purple|Pink|Ultramarine|Yellow|Titanium|Red)'
df['color'] = df['product_name'].str.extract(color_pattern, flags=re.IGNORECASE, expand=False).str.capitalize()

# Preview the newly extracted product features
df[['product_name', 'model', 'generation', 'storage', 'color']].head(10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set figure size and style
plt.figure(figsize=(8, 5))
sns.set_theme(style="whitegrid")

# Create Bar Plot
ax = sns.countplot(
    x='platform_name', 
    data=df, 
    order=df['platform_name'].value_counts().index,
    palette=['#2874F0', '#FF9900', '#000000']  # Flipkart Blue, Amazon Orange, JioMart Dark
)

# Add title and labels
plt.title('Total Orders / Listings by Platform', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('E-Commerce Platform', fontsize=12)
plt.ylabel('Number of Orders', fontsize=12)

# Annotate exact numbers on top of each bar
for p in ax.patches:
    ax.annotate(
        f'{int(p.get_height()):,}', 
        (p.get_x() + p.get_width() / 2., p.get_height()), 
        ha='center', va='bottom', 
        fontsize=11, fontweight='bold', 
        xytext=(0, 5), textcoords='offset points'
    )

plt.ylim(0, df['platform_name'].value_counts().max() * 1.15)  # Add space above top bar
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.boxplot(x='platform_name', y='price', data=df, palette='Set2')

plt.title('Price Spread and Distribution Across Platforms', fontsize=14, fontweight='bold')
plt.xlabel('Platform', fontsize=12)
plt.ylabel('Price (₹)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Statistical summary of price by platform
platform_price_summary = df.groupby('platform_name')['price'].agg(
    Listing_Count='count',
    Average_Price='mean',
    Median_Price='median',
    Min_Price='min',
    Max_Price='max'
).reset_index()

# Format numbers nicely
platform_price_summary['Average_Price'] = platform_price_summary['Average_Price'].round(2)
platform_price_summary
```

```python
# Filter out 'Unknown' state
known_states_df = df[df['state'] != 'Unknown']

# Compute Top 10 States with Order Count and Percentage Share
state_summary = known_states_df['state'].value_counts().head(10).reset_index()
state_summary.columns = ['State', 'Order_Count']

# Add Percentage Share column
state_summary['Percentage_Share'] = (state_summary['Order_Count'] / len(known_states_df) * 100).round(2)
state_summary['Percentage_Share'] = state_summary['Percentage_Share'].astype(str) + '%'

state_summary
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Prepare Top 10 States data
top_states = df[df['state'] != 'Unknown']['state'].value_counts().head(10)

plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# Create Horizontal Bar Plot
ax = sns.barplot(
    x=top_states.values, 
    y=top_states.index, 
    palette='viridis'
)

plt.title('Top 10 States by Number of iPhone Orders', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Number of Orders', fontsize=12)
plt.ylabel('State', fontsize=12)

# Annotate exact order count next to each bar
for p in ax.patches:
    width = p.get_width()
    ax.annotate(
        f'{int(width):,}', 
        (width, p.get_y() + p.get_height() / 2.), 
        ha='left', va='center', 
        fontsize=10, fontweight='bold', 
        xytext=(5, 0), textcoords='offset points'
    )

plt.xlim(0, top_states.max() * 1.15)
plt.tight_layout()
plt.show()
```

```python
# 1. Filter out 'Unknown' state
known_states = df[df['state'] != 'Unknown']

# 2. Calculate Total Purchase Revenue across all known states
total_revenue = known_states['price'].sum()

# 3. Group by state and calculate Total Purchase Value (Sum of Price)
state_purchase_df = known_states.groupby('state')['price'].sum().reset_index()
state_purchase_df.columns = ['State', 'Total_Purchase_Value']

# 4. Sort top 10 states by Total Purchase
top10_purchase_states = state_purchase_df.sort_values(by='Total_Purchase_Value', ascending=False).head(10).copy()

# 5. Format value in ₹ Crores / Lakhs
top10_purchase_states['Formatted_Value'] = top10_purchase_states['Total_Purchase_Value'].apply(
    lambda x: f"₹{x/10000000:.2f} Cr" if x >= 10000000 else f"₹{x/100000:.2f} L"
)

# 6. Add Percentage Share with respect to Total Revenue
top10_purchase_states['Percentage_Share'] = (top10_purchase_states['Total_Purchase_Value'] / total_revenue * 100).round(2).astype(str) + '%'

# Display table
top10_purchase_states
```
